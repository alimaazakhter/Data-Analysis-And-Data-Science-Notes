```python
import numpy as np
```

NumPy Indexing And Slicing of Vectors

```python
arr = np.arange(11, 21)
arr
```

```python
arr[9] #indexing returns the value at the specified index in the array.
```

```python
arr[1:5] #slicing returns a new array that contains a subset of the elements from the original array.
```

```python
arr[:5] 
```

```python
arr[:]
```

```python
arr[3:]
```

```python
arr[3::2] #start at index 3 and return every second element from that point onward.
```

NumPy Indexing and Slicing of Matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5,4]
```
